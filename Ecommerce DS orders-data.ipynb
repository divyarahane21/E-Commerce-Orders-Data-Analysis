```python
df = df.drop_duplicates()
```

```python
#10.Perform Casting columns to appropriate data types where applicable.
df.dtypes
```

```python

# 10. Perform Casting columns to appropriate data types

df['PURCHASE_TS'] = pd.to_datetime(df['PURCHASE_TS'],format='mixed',dayfirst=True,errors='coerce')

df['SHIP_TS'] = pd.to_datetime(df['SHIP_TS'],format='mixed',dayfirst=True,errors='coerce')

df.dtypes
```

```python
df.dtypes
```

```python
#  Data Visualization on the DataFrame
# 1.Design Product-wise Revenue from Orders.

# Product-wise Revenue from Orders

product_revenue = df.groupby('PRODUCT_NAME')['USD_PRICE'].sum().sort_values(ascending=False)

plt.figure(figsize=(12,6))

ax = sns.barplot(
    x=product_revenue.index,
    y=product_revenue.values
)

plt.title('Product-wise Revenue from Orders')
plt.xlabel('Product Name')
plt.ylabel('Revenue (USD)')

plt.xticks(rotation=45)

for i, value in enumerate(product_revenue.values):
    ax.text(i, value, f'{value:.2f}', ha='center', va='bottom')

plt.show()
```

```python
# Product-wise Revenue from Orders

product_revenue = df.groupby('PRODUCT_NAME')['USD_PRICE'].sum().sort_values(ascending=False)

plt.figure(figsize=(12,6))

sns.barplot(
    x=product_revenue.index,
    y=product_revenue.values
)

plt.title('Product-wise Revenue from Orders')
plt.xlabel('Product Name')
plt.ylabel('Revenue (USD)')

plt.xticks(rotation=45)

plt.show()
```

```python
#2.Design Orders by Marketing Channels.  
# Orders by Marketing Channels

channel_orders = df['MARKETING_CHANNEL'].value_counts()

plt.figure(figsize=(10,5))

ax = sns.barplot(
    x=channel_orders.index,
    y=channel_orders.values
)

plt.title('Orders by Marketing Channels')
plt.xlabel('Marketing Channel')
plt.ylabel('Number of Orders')


for i, value in enumerate(channel_orders.values):
    ax.text(i, value, str(value), ha='center', va='bottom')

plt.show()
```

```python
#3.Design Weekly Patterns in Oders Shipped (2019 - 2021).

# Filter data for 2019 to 2021
ship_data = df[df['SHIP_TS'].dt.year.between(2019, 2021)]

# Get week number
weekly_orders = ship_data.groupby(
    ship_data['SHIP_TS'].dt.isocalendar().week
).size()

# Line plot
plt.figure(figsize=(12,5))

sns.lineplot(
    x=weekly_orders.index,
    y=weekly_orders.values,
    marker='o',color='green'
)

plt.title('Weekly Patterns in Orders Shipped (2019 - 2021)')
plt.xlabel('Week Number')
plt.ylabel('Orders Shipped')

plt.xticks(range(1, 53))

plt.show()
```

```python
#4.Design Yearly Order Volume Trends.
# 4. Yearly Order Volume Trends
# Extract year and count orders
yearly_orders = df['PURCHASE_TS'].dt.year.value_counts().sort_index()

# Line Plot
plt.figure(figsize=(10,5))

sns.lineplot(
    x=yearly_orders.index,
    y=yearly_orders.values,
    marker='o'
)

plt.title('Yearly Order Volume Trends')
plt.xlabel('Year')
plt.ylabel('Number of Orders')

plt.show()
```

```python
#5.Design Orders Shipped by Year.
# 5. Orders Shipped by Year

shipped_year = df['SHIP_TS'].dt.year.value_counts().sort_index()

plt.figure(figsize=(10,5))

sns.barplot(
    x=shipped_year.index,
    y=shipped_year.values,color='violet'
)

plt.title('Orders Shipped by Year')
plt.xlabel('Year')
plt.ylabel('Orders Shipped')

plt.show()
```
